```python
# Convert TotalCharges to numeric
total_charges_numeric = pd.to_numeric(
    df["TotalCharges"].str.strip(),
    errors="coerce"
)

# Identify missing values
missing_mask = total_charges_numeric.isna()

# Validate our assumption
print("All missing values have tenure = 0:",
      (df.loc[missing_mask, "tenure"] == 0).all())

print("All missing values have Churn = No:",
      (df.loc[missing_mask, "Churn"] == "No").all())

# Create a cleaned copy for inspection
total_charges_clean = total_charges_numeric.mask(
    missing_mask & (df["tenure"] == 0),
    0
)

print("\nMissing values after handling:",
      total_charges_clean.isna().sum())

print("Number of zero TotalCharges values:",
      (total_charges_clean == 0).sum())
```

```python
plt.figure(figsize=(7, 5))

sns.countplot(data=df, x="Churn")

plt.title("Customer Churn Distribution")
plt.xlabel("Churn")
plt.ylabel("Number of Customers")

plt.show()
```

```python
# Calculate churn percentage for each contract type
contract_churn = pd.crosstab(
    df["Contract"],
    df["Churn"],
    normalize="index"
) * 100

print(contract_churn.round(2))
```

```python
contract_churn.plot(
    kind="bar",
    stacked=True,
    figsize=(8, 5)
)

plt.title("Churn Distribution by Contract Type")
plt.xlabel("Contract Type")
plt.ylabel("Percentage of Customers")
plt.xticks(rotation=0)
plt.legend(title="Churn")
plt.tight_layout()
plt.show()
```

Contract type shows a strong association with customer churn. Month-to-month customers have a substantially higher churn rate (42.71%) compared to one-year (11.27%) and two-year customers (2.83%).

```python
print("Skewness:", total_charges_clean.skew())

Q1 = total_charges_clean.quantile(0.25)
Q3 = total_charges_clean.quantile(0.75)

IQR = Q3 - Q1

lower_bound = Q1 - 1.5 * IQR
upper_bound = Q3 + 1.5 * IQR

outliers = total_charges_clean[
    (total_charges_clean < lower_bound) |
    (total_charges_clean > upper_bound)
]

print("Lower bound:", lower_bound)
print("Upper bound:", upper_bound)
print("Number of potential outliers:", len(outliers))
```

TotalCharges shows a positive skewness of 0.96, indicating a moderately right-skewed distribution. The IQR method identified no potential outliers, as all values fall within the calculated bounds. Therefore, no outlier removal is currently considered necessary for this feature.

```python
numerical_features = [
    "tenure",
    "MonthlyCharges"
]

for feature in numerical_features:
    Q1 = df[feature].quantile(0.25)
    Q3 = df[feature].quantile(0.75)
    IQR = Q3 - Q1

    lower_bound = Q1 - 1.5 * IQR
    upper_bound = Q3 + 1.5 * IQR

    outliers = df[
        (df[feature] < lower_bound) |
        (df[feature] > upper_bound)
    ]

    print(f"\n{feature}")
    print("Skewness:", round(df[feature].skew(), 3))
    print("Lower bound:", round(lower_bound, 2))
    print("Upper bound:", round(upper_bound, 2))
    print("Potential outliers:", len(outliers))
```

The numerical features were examined for skewness and potential outliers using the IQR method.

 - Tenure shows a slight positive skewness (0.24).
 - MonthlyCharges shows a slight negative skewness (-0.221).
 - TotalCharges shows a moderate positive skewness (0.963).

 No potential outliers were identified in any of the three numerical features using the IQR method. Therefore, no outlier removal is currently considered necessary.

```python
demographic_features = [
    "SeniorCitizen",
    "Partner",
    "Dependents"
]

for feature in demographic_features:
    print(f"\nChurn distribution by {feature}:")

    result = pd.crosstab(
        df[feature],
        df["Churn"],
        normalize="index"
    ) * 100

    print(result.round(2))
```

Demographic features show noticeable associations with customer churn.

 Senior citizens have a higher churn rate (41.68%) compared to non-senior customers (23.61%). Customers without a partner also show a higher churn rate (32.96%) than those with a partner (19.66%). Similarly, customers without dependents have a higher churn rate (31.28%) compared to customers with dependents (15.45%).

 These findings suggest that demographic characteristics may provide useful information for churn prediction. However, these associations do not imply causation.

```python
categorical_features = [
    "gender",
    "PhoneService",
    "PaperlessBilling",
    "MultipleLines"
]

for feature in categorical_features:
    print(f"\nChurn distribution by {feature}:")

    result = pd.crosstab(
        df[feature],
        df["Churn"],
        normalize="index"
    ) * 100

    print(result.round(2))
```

 Gender and PhoneService show very small differences in churn rates between their categories, suggesting weak individual associations with customer churn. MultipleLines also shows relatively small differences, with customers who have multiple lines experiencing a slightly higher churn rate (28.61%).

 In contrast, PaperlessBilling shows a more noticeable association with churn. Customers with paperless billing have a churn rate of 33.57%, compared to 16.33% among customers without paperless billing. This feature may provide useful information for churn prediction. However, these associations do not imply causation.

```python
service_features = [
    "OnlineBackup",
    "DeviceProtection",
    "StreamingTV",
    "StreamingMovies"
]

for feature in service_features:
    print(f"\nChurn distribution by {feature}:")

    result = pd.crosstab(
        df[feature],
        df["Churn"],
        normalize="index"
    ) * 100

    print(result.round(2))
```

OnlineBackup and DeviceProtection show noticeable associations with customer churn. Customers without OnlineBackup have a churn rate of 39.93%, compared to 21.53% among customers with the service. Similarly, customers without DeviceProtection have a churn rate of 39.13%, compared to 22.50% among those with the service.

StreamingTV and StreamingMovies show smaller differences in churn rates between customers with and without these services. Customers with no internet service have a churn rate of 7.40% across all four features.

These findings suggest that additional protection and backup services may provide useful information for churn prediction, although no causal relationship can be established from these results.

```python
features_to_compare = [
    "OnlineBackup",
    "DeviceProtection",
    "StreamingTV",
    "StreamingMovies"
]

for feature_1, feature_2 in [
    ("OnlineBackup", "DeviceProtection"),
    ("StreamingTV", "StreamingMovies")
]:
    print(f"\n{feature_1} vs {feature_2}")

    print(
        pd.crosstab(
            df[feature_1],
            df[feature_2],
            normalize="index"
        ).round(3)
    )
```

Pairwise analysis of service-related features reveals noticeable associations between OnlineBackup and DeviceProtection, as well as between StreamingTV and StreamingMovies.

Among customers with StreamingTV, 71.7% also subscribe to StreamingMovies. Similarly, 64.2% of customers without OnlineBackup also do not have DeviceProtection.

These relationships suggest that some service-related features may contain overlapping information. However, the features are not identical and may still contribute distinct information to the predictive model. Therefore, all features will initially be retained for model development.
